Numpy: -    Made in C for python
NumPy is the fundamental package for scientific computing in Python. It is a Python library that provides a multidimensional array object, various derived objects (such as masked arrays and matrices), and an assortment of routines for fast operations on arrays, including mathematical, logical, shape manipulation, sorting, selecting, I/O, discrete Fourier transforms, basic linear algebra, basic statistical operations, random simulation and much more.

At the core of the NumPy package, is the ndarray object. This encapsulates n-dimensional arrays of homogeneous(same) data types

```python
#Creating np arrays: -
import numpy as np

a = np.array([1, 2, 3])     #1d array(vector)
print(a)
print(type(a))

b = np.array([[1, 2, 3], [5, 6, 7]])    #2d array(matrix)
print(b)

c = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])      #3d array(tensor)
print(c)
```

```python
#Datatype: -    (dtype)
a = np.array([1, 2, 4], dtype='float')
print(a)

b = np.array([1, 3, 0], dtype='bool')
print(b)
```

```python
#np.arange: -
b = np.arange(1, 11)
#Works same as range funtion
print(b)
```

```python
#np.reshape(): -
a = np.arange(1, 11).reshape(5, 2)
print(a)
```

```python
#np.zeros and np.ones : -
a = np.ones((3, 4))     #Have to pass a tuple as input.
print(a)

b = np.zeros((4, 4))
print(b)

#Here we use this to initialise.
```

```python
#np.random: -
c = np.random.random((3, 4))
print(c)
```

```python
#np.linspace :- np.linearspace(full form)
a = np.linspace(-10, 10, 10)    #np.linspace(lower_bound, upper_bound, no_of_numbers_to_be _generated).
print(a)

#This linspace generates equally distanced points. All the points are linearly spaced.
```

Array attributes: -

```python
a1 = np.arange(10)  #vector
a2 = np.arange(12, dtype='float').reshape(3, 4) #matrix
a3 = np.arange(8).reshape(2, 2, 2)  #tensor
```

```python
#ndim : - To check in how many dimensions the array exists.
print(a1.ndim)
print(a2.ndim)
print(a3.ndim)
```

```python
#shape : - (gives no.of items which exist in each dimension)
print(a1.shape)     #gives length
print(a2.shape)     #gives length and wid
print(a3.shape)     #(2, 2, 2)
```

```python
#size : -   gives no.of items in the array
print(a1.size)
print(a2.size)
print(a3.size)
```

```python
#itemsize : - tells how much each item is taking space in the memory.(in bytes)
print(a1.itemsize)
print(a2.itemsize)
print(a3.itemsize)

#Here float and int carry same 8 bytes in the memory here.
```

```python
#dtype : - Tells the datatype.
print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

Array Functions: -

```python
a1 = np.random.random((3, 3))
a1 = np.round(a1*100)  
a1
```

```python
#max, min, prod, sum
print(np.max(a1))       
print(np.min(a1))
print(np.sum(a1))
print(np.prod(a1))      #product of entire array
```

```python
#If we want to find maximum in each row or column

#0 -> col and 1 -> row
print(np.min(a1, axis=1))   #minimum in each row
print(np.max(a1, axis=0))   #maximum in each column
```

```python
#Statistical function
#mean, median, std, var
print(np.mean(a1))  #mean of entire array
print(np.mean(a1, axis=1))      #mean of each row

print(np.var(a1))       #prints variance of the array.
```

```python
#Trigonometric function
print(np.sin(a1))       #sin of each element in the array
```

```python
#Dot product: -
a2 = np.arange(12).reshape(3, 4)
a3 = np.arange(12, 24).reshape(4, 3)
#Dot product resultant matrix will be 3x3 in the above case

print(np.dot(a2, a3))       #returns the resultant dot product array
```

```python
#log and exponents: -
print(np.log(a1))
print(np.exp(a1))

#log and exp of every element in the matrix.
```

```python
#round, floor and ceil
a5 = np.random.random((2, 3))*100
print(a5)

print(np.round(a5))      #This round function rounds off to nearest integer
print(np.ceil(a5))          #rounds to the greater integer
print(np.floor(a5))         #rounds to the lower integer
```

Indexing and Slicing: -

```python
a1 = np.arange(10)
a2 = np.arange(12).reshape(3, 4)
a3 = np.arange(8).reshape(2, 2, 2)
```

```python
print(a2)

#For printing 6
print(a2[1, 2])
```

```python
print(a3)

#To print 5
print(a3[1, 0, 1])        #a5[which 2x2 matrix does 5 exist, which row, which column]

#to print 3
print(a3[0, 1, 1])
```

```python
#slicing : -
#1d case(vector)
print(a1)

#print 2, 3, 4
print(a1[2:5])
```

```python
#2d array(matrix)

print(a2)
print()

#print first row with all columns
print(a2[0,:])
print()

#print 3rd column with all rows
print(a2[:,3])      # : includes all
print()

#to print 5 6 9 10 in the matrix
print(a2[1:, 1:3])
print()

#to print 0 3 8 11
print(a2[::2,::3])  #Here row jumps by 2 and column jumps by 3
print()

#to print 1 3 9 11
print(a2[0::2, 1::2])
print()

#to print 4 7
print(a2[1:-1, 0::3])
print()

```

```python
#3d array(tensor): -
a3 = np.arange(27).reshape(3, 3, 3)
print(a3)
print()

#print middle 3x3 array array
print(a3[1,:,:])
print()

#print 3 4 5
print(a3[0, 1, :])
print()

#print 10 13 16
print(a3[1,:,1])
print()

#print 0 2 18 20
print(a3[::2, 0, ::2])
```

Iterating: -

```python
#In 1d array, its same as in python

#in 2d array(matrix): - #one row is printed at a time.
for i in a2:
    print(i)
print()

#in 3d array(tensor): - #One 2d array is printed at a time
for i in a3:
    print(i)
```

```python
#To loop through each element: - np.nditer(array_name)      #To access each array.
for i in np.nditer(a3):
    print(i)
```

Stacking: - stack up two or more numpy arrays
Have to be of same dimensions.

```python
#horizontal stacking: -
a4 = np.arange(12).reshape(3, 4)
a5 = np.arange(12, 24).reshape(3, 4)

print(a4)
print()
print(a5)
print()

print(np.hstack((a4, a5)))      #Need to pass tuple of arrays as input.
```

```python
#Vertical stacking
print(np.vstack((a4, a5)))
print()

print(np.vstack((a4, a5, a4)))      #stacking multiple matrices
```

Splitting: - Opposite of stacking

```python
#horizontal splitting: - Splitting a matrix into parts with vertical cut.
print(np.hsplit(a4, 4))     #Splits into 4 parts

#at max can split on no.of columns(Here 4)
```

```python
#vertical split: - Splitting the matrix into parts with horizontal line.
print(np.vsplit(a4, 3))

#At max can split based of no.of rows(Here 3)

```
