```python
#An atm class: -
class Atm:

    def __init__(self):
        self.pin = ''
        self.__balance = 0      #This is private variable -> _Atm__balance is the new name of these.
        self.menu()

    def menu(self):
        usr_input = input("""
        Hi how can I help you?
        1. Press 1 to create pin
        2. Press 2 to change pin
        3. Press 3 to check the balance
        4. Press 4 to withdraw
        5. Anything else to exit
        """)

        if usr_input == '1':
            self.create_pin()
        elif usr_input == '2':
            self.change_pin()
        elif usr_input == '3':
            self.check_balance()
        elif usr_input == '4':
            self.withdraw()
        else:
            pass

    def create_pin(self):
        usr_pin = input('Enter your pin: ')
        self.pin = usr_pin
        usr_balance = input("Enter your current balance: ")
        self.__balance = usr_balance
        print("Pin created successfully")
        self.menu()

    def check(self):
        old_pin = input("Enter pin to verify: ")
        if old_pin == self.pin:
            return True
        else:
            print("Access denied as wrong pin entered.")
            return False
        
    def change_pin(self):
        if self.check():
            new_pin = input("Enter new pin: ")
            self.pin = new_pin
        self.menu()

    def check_balance(self):
        if self.check():
            print(f"Your current balance: {self.__balance}")
        self.menu()

    def withdraw(self):
        if self.check():
            withdraw_amt = input("Enter the amount to be withdrawn: ")

            if withdraw_amt > self.__balance:
                print("Insufficient balance")
            else:
                self.__balance -= withdraw_amt
                print("Amount withdrawn successfully")
        self.menu()

    #getter: - Get private value outside the class.
    def get_balance(self):
        return self.__balance

    #setter: - set value of private variable outside the class.
    def set_balance(self, new_bal):
        if isinstance(new_bal, int):
            self.__balance = new_bal
        else:
            print('Enter valid datatype for the balance.')
    
        
#Here balance can be accessed from outside. But it should not be accessed from outside. Hence we make it private
#self.__variable_name makes the attribute or method private
#can be accessed by _class_name__variable_name -> private variable can be accessed.

#When we try to access it using __variable_name -> This acts as a completely new attribute, not the private variable.


#To access private variables variables outside the class, we use getter and setter. This is formal way of taking private variables into access.
```

Aggregation : - (Has a relation)
One class owns the other class.
Ex: - Customer has an address. Customer class owns the address class.

Sending other class's object as input to someother class.

```python
class Customer:
    def __init__(self, name, gender, address):
        self.name=name
        self.gender = gender
        self.address = address      #A little complex entity

    def print_address(self):
        print(self.address.get_city(), self.address.pin, self.address.state)

    def edit_profile(self, new_name, new_city, new_pin, new_state):
        self.address.edit_address(new_city, new_pin, new_state)
        self.name = new_name


class Address:
    def __init__(self, city, pin, state):
        self.__city = city
        self.pin = pin
        self.state = state

    def get_city(self):
        return self.__city

    def edit_address(self, new_city, new_pin, new_state):
        self.__city = new_city
        self.pin = new_pin
        self.state = new_state

add1 = Address('Hyderabad', 502032, 'Telangana')
cust = Customer('Arun', 'Male', add1)       #here we are sending an object of other class as parameter to other class.
cust.print_address()

#You cannot fetch private attributes from the class which the other class owns. Have to use getter and setter to access private variables.

cust.edit_profile('Nikhil', 'Pala', 501011, 'Kerala')
cust.print_address()
```

Inheritence: -
Child class can access all attributes and methods of parent class. (Except private variables)
-> This helps in code reusability.
-> To follow DRY(Dont repeat yourself)

Parent cannot access child attributes or methods.

```python
class User:
    def __init__(self, name):
        self.name = name

    def login(self):
        print('Login')

class Student(User):
    def __init__(self, name, roll_no):
        super().__init__(name)
        self.roll_no = roll_no

u = User('Nikhil')
s = Student('Nithin', '2025BCS0125')
s.login()       #Here we access the method of parent class from child class
print(s.name)


#If a child class is having constructor, then first child class constructor runs. Else, the parent class class constructor runs when a child class object is created.
```

What all gets inherited : -

```python
#constructor inherit: -
class Phone:
    def __init__(self, price, brand, camera):
        print("Inside parent constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print('Buying a phone')

class SmartPhone(Phone):
    def __init__(self, os, ram):
        print('Inside child constructor')
        self.os = os
        self.ram = ram

s = SmartPhone('Android', 2)
```

-> If child has no constructor of its own, then parent's constructor is called.
-> If child has the constructor of its own, then parents constructor is not called, hence the variables in the parent constructor are not formed and we cannot use them from child constructor.
-> child cant access private members of the parent. (neither attribute nor method)

```python
#example: -
class A:
    def __init__(self):
        self.var1 = 100

    def display1(self, var1):
        #if we add the line self.var1 = var1 then output will be 200. But now self.var1 value is 100.
        print(f"class A: {self.var1}")

class B(A):
    pass

obj = B()
obj.display1(200)       #This gives the output 100 and not 200
```

Super Keyword: -

```python
class Phone:
    def __init__(self, price, brand, camera):
        print("Inside parent constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print('Buying a phone')

class SmartPhone(Phone):
    def buy(self):
        print("Buying a smartphone")
        #Syntax to call parent's buy method
        super().buy()

s = SmartPhone(20000, 'Apple', 13)
s.buy()
```

```python
#super for constructor: -
class Phone:
    def __init__(self, price, brand, camera):
        print("Inside parent constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

class SmartPhone(Phone):
    def __init__(self, price, brand, camera, os, ram):
        print('Inside child constructor')
        #syntax to call parent constructor
        super().__init__(price, brand, camera)
        self.os = os
        self.ram = ram

s = SmartPhone(20000, 'Samsung', 12, 'Android', 2)

print(s.os)
print(s.brand)
```

-> Using super we can only access methods and not attributes. (constructor is also a method)
-> super cannot be used outside the class
-> super can only be used inside child class.

Types of inheritance: -
1) Single inheritence       (child inherits from the parent)
2) Multilevel inheritence   (grandfather <- father <- child)(child can inherit grandfather too)
3) Hierachical inheritance  (father <- child1)(father <- child2)(father <- child3)(father has three children class)
4) Multiple inheritance     (parent1 <- child -> parent2)(1 child may have multiple parents)(Method resolution order is followed)
5) Hybrid inheritance       (combination of above types inheritance)

Abstraction: - (hidden)
(We can apply constraints on our lower classes)
Upper class hiding few things from the lower classes.

```python
#Abstract class (The class which contains atleast one abstract method: - We leave the class empty)
from abc import ABC, abstractmethod
class BankApp(ABC): #Class has to inherit from ABC to become abstract class
    def database(self):
        print('Connected to database')

    #abstract method
    @abstractmethod
    def security(self):
        pass
```

```python
class mobileApp(BankApp):
    def mobile_login(self):
        print('Login into mobile')

mob = mobileApp()
#Gives an error because we didnt implement the abstract method in our child class.
```

```python
class mobileApp(BankApp):
    def mobile_login(self):
        print('Login into mobile')

    def security(self):
        print('Mobile security')

mob = mobileApp()       #Now it runs perfectly
mob.security()
```
