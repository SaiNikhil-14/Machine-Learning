Encapsulation: -
To make the data private, public and protected.

```python
#Instance variable: -
class Person:
    def __init__(self, name, country):
        self.name = name
        self.country = country
        #here name and country are instance variable

#Instance variables are the variables, who have different values for different objects of the class.
p1 = Person('Arun', 'India')
p2 = Person('Ajay', 'USA')
print(p1.name)
print(p2.name)
#Here same 'name' attribute has two values as it changes for each attribute.
```

```python
#An atm class: -
class Atm:

    def __init__(self):
        self.pin = ''
        self.__balance = 0      #This is private variable -> _Atm__balance is the new name of these.
        self.menu()

    def menu(self):
        usr_input = input("""
        Hi how can I help you?
        1. Press 1 to create pin
        2. Press 2 to change pin
        3. Press 3 to check the balance
        4. Press 4 to withdraw
        5. Anything else to exit
        """)

        if usr_input == '1':
            self.create_pin()
        elif usr_input == '2':
            self.change_pin()
        elif usr_input == '3':
            self.check_balance()
        elif usr_input == '4':
            self.withdraw()
        else:
            pass

    def create_pin(self):
        usr_pin = input('Enter your pin: ')
        self.pin = usr_pin
        usr_balance = input("Enter your current balance: ")
        self.__balance = usr_balance
        print("Pin created successfully")
        self.menu()

    def check(self):
        old_pin = input("Enter pin to verify: ")
        if old_pin == self.pin:
            return True
        else:
            print("Access denied as wrong pin entered.")
            return False
        
    def change_pin(self):
        if self.check():
            new_pin = input("Enter new pin: ")
            self.pin = new_pin
        self.menu()

    def check_balance(self):
        if self.check():
            print(f"Your current balance: {self.__balance}")
        self.menu()

    def withdraw(self):
        if self.check():
            withdraw_amt = input("Enter the amount to be withdrawn: ")

            if withdraw_amt > self.__balance:
                print("Insufficient balance")
            else:
                self.__balance -= withdraw_amt
                print("Amount withdrawn successfully")
        self.menu()

    #getter: - Get private value outside the class.
    def get_balance(self):
        return self.__balance

    #setter: - set value of private variable outside the class.
    def set_balance(self, new_bal):
        if isinstance(new_bal, int):
            self.__balance = new_bal
        else:
            print('Enter valid datatype for the balance.')
    
        
#Here balance can be accessed from outside. But it should not be accessed from outside. Hence we make it private
#self.__variable_name makes the attribute or method private
#can be accessed by _class_name__variable_name -> private variable can be accessed.

#When we try to access it using __variable_name -> This acts as a completely new attribute, not the private variable.


#To access private variables variables outside the class, we use getter and setter. This is formal way of taking private variables into access.
```

Collection of objects: -

```python
class Person: 
    def __init__(self, name, gender):
        self.name = name
        self.gender = gender

p1 = Person('Nikhil', 'Male')
p2 = Person('Rahul', 'Male')
p3 = Person('Prakriti', 'Female')

#list of object
L = [p1, p2, p3]    #This is the collection of objects.

for person in L:
    print(person.name, person.gender)
```

```python
d = {'p1': p1, 'p2': p2, 'p3': p3}
for key in d:
    print(key)

for key in d:
    print(d[key].name, d[key].gender)
```

```python
class Atm:

    def __init__(self):
        self.pin = ''
        self.__balance = 0    
        self.cid = 0
        self.cid+=1
        self.menu()

    def menu(self):
        usr_input = input("""
        Hi how can I help you?
        1. Press 1 to create pin
        2. Press 2 to change pin
        3. Press 3 to check the balance
        4. Press 4 to withdraw
        5. Anything else to exit
        """)

        if usr_input == '1':
            self.create_pin()
        elif usr_input == '2':
            self.change_pin()
        elif usr_input == '3':
            self.check_balance()
        elif usr_input == '4':
            self.withdraw()
        else:
            pass

    def create_pin(self):
        usr_pin = input('Enter your pin: ')
        self.pin = usr_pin
        usr_balance = input("Enter your current balance: ")
        self.__balance = usr_balance
        print("Pin created successfully")
        self.menu()

    def check(self):
        old_pin = input("Enter pin to verify: ")
        if old_pin == self.pin:
            return True
        else:
            print("Access denied as wrong pin entered.")
            return False
        
    def change_pin(self):
        if self.check():
            new_pin = input("Enter new pin: ")
            self.pin = new_pin
        self.menu()

    def check_balance(self):
        if self.check():
            print(f"Your current balance: {self.__balance}")
        self.menu()

    def withdraw(self):
        if self.check():
            withdraw_amt = input("Enter the amount to be withdrawn: ")

            if withdraw_amt > self.__balance:
                print("Insufficient balance")
            else:
                self.__balance -= withdraw_amt
                print("Amount withdrawn successfully")
        self.menu()

    def get_balance(self):
        return self.__balance

    def set_balance(self, new_bal):
        if isinstance(new_bal, int):
            self.__balance = new_bal
        else:
            print('Enter valid datatype for the balance.')
```

```python
c1 = Atm()
c2 = Atm()
c3 = Atm()

print(c1.cid)
print(c2.cid)
print(c3.cid)
#Here all the three print the same ans 1. As here the cid acts as an instance variable and acts independent for every object.
```

```python
class Atm:
    count = 1
    def __init__(self):
        self.pin = ''
        self.__balance = 0    
        self.menu()
        self.cid = Atm.count
        Atm.count+=1

    def menu(self):
        usr_input = input("""
        Hi how can I help you?
        1. Press 1 to create pin
        2. Press 2 to change pin
        3. Press 3 to check the balance
        4. Press 4 to withdraw
        5. Anything else to exit
        """)

        if usr_input == '1':
            self.create_pin()
        elif usr_input == '2':
            self.change_pin()
        elif usr_input == '3':
            self.check_balance()
        elif usr_input == '4':
            self.withdraw()
        else:
            pass

    def create_pin(self):
        usr_pin = input('Enter your pin: ')
        self.pin = usr_pin
        usr_balance = input("Enter your current balance: ")
        self.__balance = usr_balance
        print("Pin created successfully")
        self.menu()

    def check(self):
        old_pin = input("Enter pin to verify: ")
        if old_pin == self.pin:
            return True
        else:
            print("Access denied as wrong pin entered.")
            return False
        
    def change_pin(self):
        if self.check():
            new_pin = input("Enter new pin: ")
            self.pin = new_pin
        self.menu()

    def check_balance(self):
        if self.check():
            print(f"Your current balance: {self.__balance}")
        self.menu()

    def withdraw(self):
        if self.check():
            withdraw_amt = input("Enter the amount to be withdrawn: ")

            if withdraw_amt > self.__balance:
                print("Insufficient balance")
            else:
                self.__balance -= withdraw_amt
                print("Amount withdrawn successfully")
        self.menu()

    def get_balance(self):
        return self.__balance

    def set_balance(self, new_bal):
        if isinstance(new_bal, int):
            self.__balance = new_bal
        else:
            print('Enter valid datatype for the balance.')

    #utility method(No need of an object to work)
    @staticmethod
    def method_name():
        print("Hi, I am an static method")
```

```python
c1 = Atm()
c2 = Atm()
c3 = Atm()

print(c1.cid)
print(c2.cid)
print(c3.cid)
#Now here cid is updated by the class variable named counter which only depends on the class and not on every individual object.
#Hence here the values change


#  Atm.count = 'hehe'           #This works
#Here we can change the class variable from outside as well if it is not private variable.

Atm.method_name()       #This is how we access static methods.
```
