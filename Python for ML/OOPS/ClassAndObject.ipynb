```python
#An atm class: -
class Atm:

    #constructor(special function) : - Whenever an object of the class is created, constructor gets called automatically. No need to call it explicitly.
    def __init__(self):
        #Here self is the object itself.
        self.pin = ''
        self.balance = 0
        self.menu()

    def menu(self):
        usr_input = input("""
        Hi how can I help you?
        1. Press 1 to create pin
        2. Press 2 to change pin
        3. Press 3 to check the balance
        4. Press 4 to withdraw
        5. Anything else to exit
        """)

        if usr_input == '1':
            self.create_pin()
        elif usr_input == '2':
            self.change_pin()
        elif usr_input == '3':
            self.check_balance()
        elif usr_input == '4':
            self.withdraw()
        else:
            pass

    def create_pin(self):
        usr_pin = input('Enter your pin: ')
        self.pin = usr_pin
        usr_balance = input("Enter your current balance: ")
        self.balance = usr_balance
        print("Pin created successfully")
        self.menu()

    def check(self):
        old_pin = input("Enter pin to verify: ")
        if old_pin == self.pin:
            return True
        else:
            print("Access denied as wrong pin entered.")
            return False
        
    def change_pin(self):
        if self.check():
            new_pin = input("Enter new pin: ")
            self.pin = new_pin
        self.menu()

    def check_balance(self):
        if self.check():
            print(f"Your current balance: {self.balance}")
        self.menu()

    def withdraw(self):
        if self.check():
            withdraw_amt = input("Enter the amount to be withdrawn: ")

            if withdraw_amt > self.balance:
                print("Insufficient balance")
            else:
                self.balance -= withdraw_amt
                print("Amount withdrawn successfully")
        self.menu()
        

```

```python
obj = Atm()     #Here obj is the object of Atm class
print(type(obj))
```

Fraction datatype

```python
class Fraction:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    #__str__() is another dunder method which return the value in it whenever we print the object of the class.
    def __str__(self):
        return "{self.a}/{self.b}"

    
    #adding two fractions
    #here self is one object and other is another object of fraction class
    #here __add__() is another magic method(activates when two objects are have + in between)
    def __add__(self, other):
        print(f"{(self.a*other.b)+(self.b*other.a)}"/{self.b*other.b})

    #another dunder method
    def __sub__(self, other):
         print(f"{(self.a*other.b)-(self.b*other.a)}"/{self.b*other.b})

    #magic method when * is seen 
    def __mul__(self, other):
        print(f"{self.a*other.a}/{self.b*other.b}")

    #magic method when / is seen 
    def __truediv__(self, other):
        print(f"{self.a*other.b}/{self.b*other.a}")

    def convert_to_dec(self):
        return self.a/self.b
```

```python
fr1 = Fraction(3, 4)
fr2 = Fraction(4, 3)
print(fr1 * fr2)
```

Functionalities: - 
1. A user can create and view 2D coordinates
2. A user can find out distance between two coordinates
3. A user can find the distance of the object from the origin 
4. A user can check if the point lies on the given line
5. A user can find the distance between a given 2D point and a given line 

```python
#A line class
class line: 
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def distance_from_point(self, point):
        print((self.a*point.x + self.b*point.y + self.c)/((self.a ** 2 + self.b ** 2) ** 0.5))

#A point class
class point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"
    
    def euclidean_dist(self, other):
        print(((self.x - other.x)**2 + (self.y - other.y)**2) ** 0.5)

    def dist_from_origin(self):
        self.euclidean_dist(point(0, 0))
        #calling eucliean_dist with (0, 0) as the other object

    def point_on_line(self, line):
        if line.a*self.x + line.b*self.y + line.c == 0:
            print("Point lies on the line.")
        else:
            print("Point does not lie on the line.")

    
```

Reference variable: -
Reference variables hold objects.
We can create objects without reference variable as well.
An object can have multiple reference variables.
Assigning new reference variable to an existing object does not create a new object.  

```python
#object without reference: -
class Person: 
    def __init__(self):
        self.name = 'Nikhil'
        self.gender = 'Male'

Person()        #This creates an object of Person class
#p = Person(), here p is the variable which holds the Person class object  
#Here p is the reference variable
```

```python
p = Person()
q = p

#here both p and q point to the same object
print(id(p))
print(id(q))
#Here p and q are reference variable
```

```python
#we can change attribute calue using second object
print(p.name)
print(q.name)

q.name = 'Sai'

print(p.name)
print(q.name)

#Here both the values change as q referes to the same object as p.
```

Pass by reference: -

```python
class Person:
    def __init__(self, name, gender):
        self.name = name
        self.gender = gender

def greet(person):
    print(f'Hi, I am {person.name}. I am a {person.gender}')
#Here the function is outside the class hence it is a function.

p = Person('Nikhil', 'Male')
greet(p)    #This is pass by reference, as we passed a reference of an object of the class(we passed the address of the object).
```
