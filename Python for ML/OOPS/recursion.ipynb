In recursion we use stack for the previous step. Stack data structure is used.

```python
#Doing a*b
def mult(a, b):
    if b==1:
        return a
    return a+mult(a, b-1)

print(mult(5, 9))
```

```python
#n factorial
def fact(n):
    if n==1:
        return 1
    return n*fact(n-1)

print(fact(4))
```

```python
#Check whether given string is palindrome.      (Hard for me)
def check(s):
    if len(s) <= 1:
        return True
    if s[0] == s[-1]:
        return check(s[1:len(s)-1])
    else:
        return False

print(check('madam'))
print(check('racecar'))
print(check('sir'))
print(check('malayalam'))
print(check('abba'))
```
