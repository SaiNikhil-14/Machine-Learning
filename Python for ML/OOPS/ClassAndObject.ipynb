```python
#An atm class: -
class Atm:

    #constructor(special function) : - Whenever an object of the class is created, constructor gets called automatically. No need to call it explicitly.
    def __init__(self):
        #Here self is the object itself.
        self.pin = ''
        self.balance = 0
        self.menu()

    def menu(self):
        usr_input = input("""
        Hi how can I help you?
        1. Press 1 to create pin
        2. Press 2 to change pin
        3. Press 3 to check the balance
        4. Press 4 to withdraw
        5. Anything else to exit
        """)

        if usr_input == '1':
            self.create_pin()
        elif usr_input == '2':
            self.change_pin()
        elif usr_input == '3':
            self.check_balance()
        elif usr_input == '4':
            self.withdraw()
        else:
            pass

    def create_pin(self):
        usr_pin = input('Enter your pin: ')
        self.pin = usr_pin
        usr_balance = input("Enter your current balance: ")
        self.balance = usr_balance
        print("Pin created successfully")
        self.menu()

    def check(self):
        old_pin = input("Enter pin to verify: ")
        if old_pin == self.pin:
            return True
        else:
            print("Access denied as wrong pin entered.")
            return False
        
    def change_pin(self):
        if self.check():
            new_pin = input("Enter new pin: ")
            self.pin = new_pin
        self.menu()

    def check_balance(self):
        if self.check():
            print(f"Your current balance: {self.balance}")
        self.menu()

    def withdraw(self):
        if self.check():
            withdraw_amt = input("Enter the amount to be withdrawn: ")

            if withdraw_amt > self.balance:
                print("Insufficient balance")
            else:
                self.balance -= withdraw_amt
                print("Amount withdrawn successfully")
        self.menu()
        

```

```python
obj = Atm()     #Here obj is the object of Atm class
print(type(obj))
```

Fraction datatype

```python
class Fraction:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    #__str__() is another dunder method which return the value in it whenever we print the object of the class.
    def __str__(self):
        return "{self.a}/{self.b}"

    
    #adding two fractions
    #here self is one object and other is another object of fraction class
    #here __add__() is another magic method(activates when two objects are have + in between)
    def __add__(self, other):
        print(f"{(self.a*other.b)+(self.b*other.a)}"/{self.b*other.b})

    #another dunder method
    def __sub__(self, other):
         print(f"{(self.a*other.b)-(self.b*other.a)}"/{self.b*other.b})

    #magic method when * is seen 
    def __mul__(self, other):
        print(f"{self.a*other.a}/{self.b*other.b}")

    #magic method when / is seen 
    def __truediv__(self, other):
        print(f"{self.a*other.b}/{self.b*other.a}")

    def convert_to_dec(self):
        return self.a/self.b
```

```python
fr1 = Fraction(3, 4)
fr2 = Fraction(4, 3)
print(fr1 * fr2)
```
