In recursion we use stack for the previous step. Stack data structure is used.

```python
#Doing a*b
def mult(a, b):
    if b==1:
        return a
    return a+mult(a, b-1)

print(mult(5, 9))
```

```python
#n factorial
def fact(n):
    if n==1:
        return 1
    return n*fact(n-1)

print(fact(4))
```

```python
#Check whether given string is palindrome.      (Hard for me)
def check(s):
    if len(s) <= 1:
        return True
    if s[0] == s[-1]:
        return check(s[1:len(s)-1])
    else:
        return False

print(check('madam'))
print(check('racecar'))
print(check('sir'))
print(check('malayalam'))
print(check('abba'))
```

```python
#Rabbit problem: -  #This is fibonacci series
def fibo(n):
    if n==0 or n==1:
        return 1
    return fibo(n-1)+fibo(n-2)

print(fibo(5))
print(fibo(4))



#This code is highly inefficient due to calculating the same part multiple times. Here DP(Dynamic programming) helps in reducing the time.
#Time complexity is O(2^n)
```

```python
#Fibo using dp
def fibo(n, d):
    if n in d:
        return d[n]
    else:
        d[n] = fibo(n-1, d) + fibo(n-2, d)
        return d[n]

d = {0:1, 1:1}
print(fibo(50, d))

#This is very very fast compared to the above code. This is more efficient.
```
