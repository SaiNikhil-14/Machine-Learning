np.sort: - Returns sorted numpy array

Returns numpy array

```python
a = np.random.randint(1, 100, 15)
print(a)
print(np.sort(a))
```

```python
b = np.random.randint(1, 100, 24).reshape(6, 4)
print(b)
print()
print(np.sort(b))       #Sorts each row
print()
print(np.sort(b, axis=0))       #sorts according to column
```

np.concatenate() : -        similar to vstack and hstack

```python
c = np.arange(6).reshape(2, 3)
d = np.arange(6, 12).reshape(2, 3)

print(c)
print(d)
```

```python
print(np.concatenate((c, d)))       #for concatenating top and bottom
print()
print(np.concatenate((c, d), axis=1))       #For concatenating side by side
```

np.expand_dims(): - Used to expand dimensions of an array.

```python
print(a)
print(a.shape)
print()

np.expand_dims(a, axis=0)
print(np.expand_dims(a, axis=0))
print(np.expand_dims(a, axis=0).shape)      #Becomes 2d
```

np.where(): - Returns the indices of the elements in an array where the given condition is satisfied.

Very powerful function and useful.

```python
print(a)

np.where(a>50)      #Returns all the index positions of all elements which are greater than 50.
```

```python
#replace all the values greater than 50 with 0
#np.where(condition, true, false)   if true what to do and if false what to do
np.where(a>50, 0, a)
```

np.argmax(): -
Returns indices of the max element in the particular axis.

```python
print(a)
print()
np.argmax(a)        #return the index of the max value element.
```

```python
print(b)
print()
print(np.argmin(b, axis=0))     #Tells min element index in each column
print(np.argmin(b, axis=1))     #tells min element index in each row
```

np.cumsum(): - Used to calculate the cumulative sum of the elements of the given array over a given axis.
Same as prefix sum.

```python
print(a)
print()
print(np.cumsum(a))     #returns the array of cumulative sum(prefix sum).
```

```python
print(b)
print()
print(np.cumsum(b, axis=0))     #cumulative sum of the column.
print()
print(np.cumsum(b, axis=1))     #cumulative sum of the row.
```

np.cumprod(): - Same as the prefix product.
Returns the cumulative product array.

```python
print(b)
print()
print(np.cumprod(b, axis=0))     #cumulative sum of the column.
print()
print(np.cumprod(b, axis=1))     #cumulative sum of the row.
```

np.percentile(): - used to compute the nth percentile of the given data(array) along the given axis.

```python
print(a)
print()
print(np.percentile(a, 100))        #The 100 percentile number(max number in the array)
print(np.percentile(a, 80))
print(np.percentile(a, 50))     #The 50th percentile will the median of the given array
print(np.percentile(a, 0))      #The 0 percentile will be the min number in the array
```

np.corrcoef: -
Returns pearson product-moment correlation coefficients.(Exists between -1 to 1)

```python
salary = np.array([20000, 40000, 25000, 35000, 60000])
experience = np.array([1, 3, 2, 4, 2])

np.corrcoef(salary, experience)
#The given matrix is [[salary-salary, salary-experience], [experience-salary, experience-experience]] correlation
#If correlation is 1 then +ve correation, -1 then -ve correlation and if 0 then no correlation.
```

SET functions: -

```python
m = np.array([1, 2, 3, 4, 5])
n = np.array([3, 4, 5, 6, 7])
```

```python
#np.union1d()
np.union1d(m, n)
```

```python
#np.intersect1d()
np.intersect1d(m, n)
```

```python
#np.setdiff1d()
print(np.setdiff1d(m, n))       #m-n
print(np.setdiff1d(n, m))       #n-m
```

```python
#np.setxor1d()
np.setxor1d(m, n)       #Removes the common elements
```

```python
#np.in1d()
np.in1d(m, 1)
np.in1d(m, 10)
```
