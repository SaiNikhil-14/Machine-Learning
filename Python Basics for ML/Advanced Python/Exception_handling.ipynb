There are 2 stages where error may take place in a program: -
1) During Compilation(Syntax error)(Error raised by compiler/interpreter)
2) During execution(Exceptions)

Syntax errors: -

```python
#1)If we leave the semicolons, brackets etc : - SyntaxError: invalid syntax

#2)If we forget to leave indentation: - IndentationError: expected an indented block

#3)IndexError

#4)ModuleNotFoundError

#5)KeyError : - When a key is not found

#6)TypeError : - When we apply an operation or function to an object of inappropriate type.     Ex: 1 + 'a'

#7)ValueError : - Thrown when a function's argument is of inappropriate type        Ex: int('a')

#8)NameError : - Thrown when an object could not be found (Using an uninitialised object)

#9) AttributeError: - Thrown when wrong function is used for wrong datatype  ex: L = [1, 2, 3] and we use L.upper(). This gives attribute error.
```

Exceptions: -


We have to hide the errors for security and user experience.

```python
#Try except block
#Try : - The block of code which we think may cause error
#except : - The block of code to be executed if Try block faces error
```

```python
#Try except: -
try:
    with open('sample.txt', r) as f:
        print(f.read())
except:
    print("File not found")

#Here instead of a long message, we simply catched the error and showed a simplified image.
```

```python
#Catching specific exception: -
try:
    f = open('sample.txt', 'r')
    print(f.read())
    print(m)
except:
    print('Some error occured')


#This is a more generalised error. Here we dont know what error occured. As there are two errors. Hence we should catch specific errors and show specific code if error occurs.
```

```python
try:
    m = 10
    f = open('Sample.txt', 'r')
    print(f.read())
    print(m)
    print(5/2)
    L = [1, 2, 3,4]
    L[100]

except FileNotFoundError:
    print('File not found')

except NameError:
    print('Variable not defined.')

except ZeroDivisionError:
    print('Cannot divide by zero')

except Exception as e:
    print(e)
    #This is to catch something which we didnt expect. (Generic messages)
    #This should always be at last

#This specifically catches the error ans shows custom message.
```

```python
#else block: 
try:
    f = open('Sample.txt', 'r')
except FileNotFoundError:
    print("The file is no present")
except Exception as e:      #Generic error
    print('Some error is occurring')
else:       #Used if no error found in try and then we want the try block code to do. This else block should not face any error!!!
    print(f.read())
```

```python
#finally: - This block of code executes in any case. Even if exception is found or not
try:
    f = open('Sample.txt', 'r')
except FileNotFoundError:
    print("The file is no present")
except Exception as e:      #Generic error
    print('Some error is occurring')
else:     
    print(f.read())
finally:
    print("This piece of code runs in any case")
```

```python
#If error found: -   try -> exception -> finally
#If error not found: -   try -> else -> finally

#Finally runs in all cases
#Generally in finally we close stuff like closing socket, closing database etc.
```

```python
#raise - exception: -
#We can manually raise exceptions using raise

raise NameError("Hi, I am an error.")
```

```python
#Benefits : - 
class Bank:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount<0:
            raise Exception("Amount to be withdrawn cannot be less than zero.")
        if amount>self.balance:
            raise Exception("Insufficient balance")
        self.balance-=amount

obj = Bank(10000)
try:
    obj.withdraw(-50000)
except Exception as e:
    print(e)        
else:
    print(obj.balance)
```

Creating custom exceptions: -

```python
#To create our custom exception class, we have to inherit Exception class to raise the error.
class MyException(Exception):
    def __init__(self, message):
        print(message)

class Bank:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount<0:
            raise MyException("Amount to be withdrawn cannot be less than zero.")
        if amount>self.balance:
            raise MyException("Insufficient balance")
        self.balance-=amount

obj = Bank(10000)
try:
    obj.withdraw(-50000)
except MyException as e:
    pass
else:
    print(obj.balance)

#This is our custom exception. We can check and raise our custom errors.
```

```python
#The above custom error classes, give us full control.
```

```python
class SecurityError(Exception):
    def __init__(self, message):
        print(message)

    def logout(self):
        print("Logout from all devices")

class Google:
    def __init__(self, name, email, password, device):
        self.name = name
        self.email = email
        self.password = password
        self.device = device

    def login(self, email, password, device):
        if device != self.device:
            raise SecurityError('Logged in from a new device!')
        if email == self.email and password == self.password:
            print("Welcome")
        else:
            print("Login Error")

obj = Google('Nikhil', 'Nikhil@gmail.com', '1234', 'Windows')
try:
    obj.login('Nikhil@gmail.com', '1234', 'Mac')
except SecurityError as e:
    e.logout()
else:
    print(obj.name)
finally:
    print('Database connection closed.')
```
